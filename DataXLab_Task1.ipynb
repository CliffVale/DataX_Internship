```python
##Import and Load
import pandas as pd

# Load the raw dataset
df = pd.read_csv('KaggleV2-May-2016.csv')

# View the first 5 rows to see what we are working with
df.head()
```

```python
##Standardize Column Headers
# Clean column names
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('-', '_')

# Verify the new headers
print(df.columns)
```

```python
##Handle Missing Values
# Identify missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Drop rows with missing values 
df.dropna(inplace=True)

# Note: If you wanted to fill them instead, you would use df.fillna(value)
```

```python
##Remove Duplicates
# Check how many rows we have before dropping
rows_before = len(df)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

rows_after = len(df)
print(f"Dropped {rows_before - rows_after} duplicate rows.")
```

```python
##Standardize Text Values
# Standardize the 'gender' column
if 'gender' in df.columns:
    df['gender'] = df['gender'].str.upper().str.strip()
    
# Check the unique values to ensure they are clean
print(df['gender'].unique())
```

```python
##Fix Data Types and Dates
# Convert scheduling dates to proper datetime objects
if 'scheduledday' in df.columns:
    df['scheduledday'] = pd.to_datetime(df['scheduledday']).dt.normalize()

if 'appointmentday' in df.columns:
    df['appointmentday'] = pd.to_datetime(df['appointmentday']).dt.normalize()

# Ensure 'age' is an integer
if 'age' in df.columns:
    df['age'] = df['age'].astype(int)

# Verify the final data types
print(df.dtypes)
```

```python
##Export the Cleaned Dataset
# Save the cleaned dataset without the index column
df.to_csv('cleaned_medical_appointments.csv', index=False)
print("Data cleaning complete and saved successfully!")
```
